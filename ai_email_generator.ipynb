```python
%%writefile app.py

import os
import streamlit as st
from groq import Groq


# =============================
# Configuration
# =============================

st.set_page_config(
    page_title="AI Email Generator",
    page_icon="✉️",
    layout="centered"
)


# =============================
# Groq API
# =============================

api_key = os.getenv("GROQ_API_KEY")

if not api_key:
    st.error("GROQ_API_KEY is not configured.")
    st.stop()

client = Groq(api_key=api_key)


# =============================
# Generate Email
# =============================

def generate_email(
    email_type,
    recipient,
    purpose,
    tone,
    length
):

    prompt = f"""
You are an AI email writing assistant.

Write a {tone.lower()} email.

Email type:
{email_type}

Recipient:
{recipient}

Purpose:
{purpose}

Length:
{length}

Requirements:
- Write clear and natural English.
- Do not use unnecessary fancy words.
- Keep the email realistic and human.
- Include a suitable subject line.
- Do not invent facts.
- Return only the subject and email body.
"""

    response = client.chat.completions.create(
        model="openai/gpt-oss-20b",
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ],
        temperature=0.7,
    )

    return response.choices[0].message.content


# =============================
# UI
# =============================

st.title("AI Email Generator")

st.write(
    "Generate professional emails in seconds using AI."
)


email_type = st.selectbox(
    "Email Type",
    [
        "Professional",
        "Job Application",
        "Follow-up",
        "Cold Email",
        "Professor",
        "Thank You",
        "Apology",
        "Custom"
    ]
)


recipient = st.text_input(
    "Who are you writing to?",
    placeholder="e.g. Professor, recruiter, client..."
)


purpose = st.text_area(
    "What do you want to say?",
    placeholder="Explain what you want the email to communicate..."
)


tone = st.selectbox(
    "Tone",
    [
        "Professional",
        "Friendly",
        "Formal",
        "Concise"
    ]
)


length = st.selectbox(
    "Length",
    [
        "Short",
        "Medium",
        "Detailed"
    ]
)


if st.button("Generate Email"):

    if not recipient or not purpose:

        st.warning(
            "Please provide the recipient and purpose."
        )

    else:

        with st.spinner("Writing your email..."):

            try:

                email = generate_email(
                    email_type,
                    recipient,
                    purpose,
                    tone,
                    length
                )

                st.subheader("Generated Email")

                st.text_area(
                    "Your email",
                    email,
                    height=350
                )

            except Exception as e:

                st.error(
                    "Something went wrong while generating the email."
                )

                st.exception(e)
```
